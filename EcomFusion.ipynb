```python
import pandas as pd
import numpy as np
```

```python
orders = pd.read_csv("Orders.csv")
customers = pd.read_csv("Customers.csv")
products = pd.read_csv("Products.csv")
```

```python
print("Orders Dataset:")
display(orders.head())

print("Customers Dataset:")
display(customers.head())

print("Products Dataset:")
display(products.head())
```

```python
print("Orders Shape:", orders.shape)
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)
```

```python
print("Orders Columns:")
print(orders.columns.tolist())

print("\nCustomers Columns:")
print(customers.columns.tolist())

print("\nProducts Columns:")
print(products.columns.tolist())
```

```python
print("Orders Data Types:")
print(orders.dtypes)

print("\nCustomers Data Types:")
print(customers.dtypes)

print("\nProducts Data Types:")
print(products.dtypes)
```

```python
print("Missing Values in Orders:")
print(orders.isnull().sum())

print("\nMissing Values in Customers:")
print(customers.isnull().sum())

print("\nMissing Values in Products:")
print(products.isnull().sum())
```

```python
orders["Order_Date"] = pd.to_datetime(
    orders["Order_Date"],
    errors="coerce"
)
```

```python
print(orders["Order_Date"].dtype)
```

```python
orders["Order_Year"] = orders["Order_Date"].dt.year
```

```python
orders["Order_Month"] = orders["Order_Date"].dt.month
```

```python
orders["Month_Name"] = orders["Order_Date"].dt.month_name()
```

```python
orders["Order_Day"] = orders["Order_Date"].dt.day
```

```python
orders["Day_of_Week"] = orders["Order_Date"].dt.day_name()
```

```python
display(
    orders[
        [
            "Order_Date",
            "Order_Year",
            "Order_Month",
            "Month_Name",
            "Order_Day",
            "Day_of_Week"
        ]
    ].head(10)
)
```

```python
def quantity_category(quantity):
    if quantity <= 2:
        return "Low"
    elif quantity <= 4:
        return "Medium"
    else:
        return "High"
```

```python
orders["Quantity_Category"] = orders["Quantity"].apply(
    quantity_category
)
```

```python
display(
    orders[
        ["Quantity", "Quantity_Category"]
    ].head(10)
)
```

```python
processed_df = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)
```

```python
display(processed_df.head())
```

```python
processed_df = pd.merge(
    processed_df,
    products,
    on="Product_ID",
    how="left"
)
```

```python
display(processed_df.head())
```

```python
processed_df["Total_Amount"] = processed_df.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)
```

```python
display(
    processed_df[
        [
            "Product_ID",
            "Product_Name",
            "Quantity",
            "Unit_Price",
            "Total_Amount"
        ]
    ].head(10)
)
```

```python
orders_part1 = orders.iloc[:60]
orders_part2 = orders.iloc[60:]
```

```python
orders_combined = pd.concat(
    [orders_part1, orders_part2],
    ignore_index=True
)
```

```python
print("Original Orders Shape:", orders.shape)
print("Orders after concat:", orders_combined.shape)
```

```python
final_columns = [
    "Order_ID",
    "Order_Date",
    "Order_Year",
    "Order_Month",
    "Month_Name",
    "Order_Day",
    "Day_of_Week",
    "Customer_ID",
    "Customer_Name",
    "City",
    "Region",
    "Membership_Type",
    "Product_ID",
    "Product_Name",
    "Category",
    "Brand",
    "Unit_Price",
    "Quantity",
    "Quantity_Category",
    "Total_Amount",
    "Payment_Method",
    "Order_Status"
]
```

```python
final_df = processed_df[final_columns].copy()
```

```python
display(final_df.head(10))
```

```python
display(final_df.sample(10, random_state=42))
```

```python
print("Final Dataset Shape:", final_df.shape)
```

```python
final_df.info()
```

```python
print("Missing Values:")
print(final_df.isnull().sum())
```

```python
duplicate_count = final_df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

```python
final_df = final_df.drop_duplicates().reset_index(drop=True)
```

```python
print("=" * 60)
print("PROCESSED E-COMMERCE DATASET SUMMARY")
print("=" * 60)

print(f"Number of Rows    : {final_df.shape[0]}")
print(f"Number of Columns : {final_df.shape[1]}")

print("\nFeatures:")
for column in final_df.columns:
    print("-", column)

print("\nData Types:")
print(final_df.dtypes)

print("\nMissing Values:")
print(final_df.isnull().sum())

print("\nDuplicate Rows:")
print(final_df.duplicated().sum())
```

```python
output_file = "Processed_Ecommerce_Dataset.csv"

final_df.to_csv(
    output_file,
    index=False
)

print(f"\nDataset successfully exported as '{output_file}'")
```
